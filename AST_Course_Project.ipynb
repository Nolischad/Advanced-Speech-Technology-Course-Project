```python
import speech_recognition as sr
from gtts import gTTS
from IPython.display import Audio
import os
import subprocess

responses = {
    "what is your return policy": "Free replacement for up to 7 days.",
    "do you offer refunds": "Full refund if product is damaged"
}

def convert_opus_to_wav(input_path, output_path):
    command = ["ffmpeg", "-i", input_path, output_path, "-y"]
    subprocess.run(command, stdout=subprocess.PIPE, stderr=subprocess.PIPE)

def transcribe_audio_google(audio_path):
    recognizer = sr.Recognizer()
    with sr.AudioFile(audio_path) as source:
        audio_data = recognizer.record(source)
    text = recognizer.recognize_google(audio_data)
    print("User said:", text)
    return text.lower()

def get_bot_response(user_input):
    return responses.get(user_input, "Sorry, I don't understand that.")

def text_to_speech(text):
    print("Bot says:", text)
    tts = gTTS(text=text, lang="en")
    tts.save("response.mp3")
    return Audio("response.mp3", autoplay=False)

audio_file_path = "/content/Chandhana refund offers.opus"
converted_audio_path = "/content/converted_audio.wav"
convert_opus_to_wav(audio_file_path, converted_audio_path)

user_text = transcribe_audio_google(converted_audio_path)
bot_response = get_bot_response(user_text)
bot_audio = text_to_speech(bot_response)
user_audio = Audio(converted_audio_path, autoplay=False)

print("Playing user voice note:")
display(user_audio)
print("Playing bot response:")
display(bot_audio)
```
